# Reward shaping ablation

Grid over (`w_format`, `w_length`) with everything else held fixed.
Two axes:
- reward hacking pressure (how much the model games shaping)
- final eval pass@1

The sweet spot from the Mar 22 debug: `w_format=0.1`, `w_length=0.005`.

```python
import json
import pandas as pd

runs = [
    {'w_format': 0.0,  'w_length': 0.0,   'reward_final': 0.68, 'gsm8k_after': 0.61, 'note': 'no shaping, format degrades'},
    {'w_format': 0.1,  'w_length': 0.0,   'reward_final': 0.78, 'gsm8k_after': 0.72, 'note': 'lucky-guess hack visible'},
    {'w_format': 0.1,  'w_length': 0.005, 'reward_final': 0.72, 'gsm8k_after': 0.73, 'note': 'stable, best eval'},
    {'w_format': 0.1,  'w_length': 0.02,  'reward_final': 0.55, 'gsm8k_after': 0.65, 'note': 'over-penalized, think block truncated'},
    {'w_format': 0.3,  'w_length': 0.005, 'reward_final': 0.81, 'gsm8k_after': 0.70, 'note': 'format-farming, eval flat'},
]
pd.DataFrame(runs)
```

```python
import matplotlib.pyplot as plt
df = pd.DataFrame(runs)
fig, ax = plt.subplots()
ax.scatter(df['reward_final'], df['gsm8k_after'])
for _, r in df.iterrows():
    ax.annotate(f"f={r['w_format']}, l={r['w_length']}", (r['reward_final'], r['gsm8k_after']), fontsize=8)
ax.set_xlabel('final training reward'); ax.set_ylabel('gsm8k eval pass@1')
ax.set_title('training reward vs actual eval')
```
