# Analyze rollouts

Load a per-benchmark JSONL dump from `eval/run_eval.py`, look at:
- reward distribution
- pass rate per subject / level (MATH)
- length distribution of pass vs fail
- top 10 failure modes by verifier `reason`

```python
import json
from collections import Counter
import pandas as pd

rows = [json.loads(l) for l in open('../benchmarks/runs/latest/after/math.jsonl')]
df = pd.DataFrame(rows)
df.head()
```

```python
df['ok'].value_counts(normalize=True)
```

```python
df['resp_len'] = df['response'].str.len()
df.groupby('ok')['resp_len'].describe()
```

```python
Counter(df.loc[~df['ok'], 'reason']).most_common(10)
```

```python
df[['subject', 'level']] = df['task_id'].str.split('-', expand=True)
df.groupby(['subject', 'level'])['ok'].mean().unstack('level')
```
